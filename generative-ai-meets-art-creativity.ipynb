# 5. What really is generative AI ? <a class='anchor' id='latex'></a> 📚[↑](#top)


###### Image courtesy of https://www.cloudskillsboost.google




Deep learning which Gen AI is founded on, is very powerful, it uses artificial neural networks, this allows them to process more complex patterns. See, artificial neural networks are inspired by the human brain this  means they are made up of many interconnected nodes or neurons that can learn to perform tasks by processing data and making predictions. Deep learning models typically have many layers of neurons, which allows them to learn more complex patterns than traditional machine learning models. And the beauty of it is that neural networks can use both labeled and unlabeled data.

This is called **semi-supervised learning**, this is a game-changer, a neural network is trained on a small amount of labeled data and a large amount of unlabeled data, the meat and potatoes of Gen AI, the labeled data then helps the neural network to learn the basic concepts of the task while the unlabeled data helps the neural network to generalize to new examples. Now this is where **generative AI fits into this AI discipline.** 

We have seen how Gen AI is a subset of deep learning, which means it uses artificial neural networks, can process both labeled and unlabeled data using supervised, unsupervised, and semi-supervised methods.**Large language models** or **LLMS** are also a subset of deep learning.LLMs and Gen AI intersect and are both a part of deep learning, LLms are large general-purpose language models, this models can be **pre-trained** and then **fine-tuned** for a specific purpose.

#### What is pre-training and fine-tuning?
LLMs are trained to solve common language problems like text generation or text classification in many industries which  only few companies can afford to do because of how expensive it is, luckily for rest of us, the models can  be tailored to solve specific problems in different fields such as finance or retail using a small size of that specific field dataset. This is a dream come true! Let us talk about GPT for example, OpenAI released GPT-4 in March 2023, GPT is a encoder **transformer** only model, We have parameters we can tweak when using the OpenAI API to get more customized results.

Now deep learning models, or machine learning models in general, can be divided into two types; 

**Discriminative Model**

A discriminative model is a type of model that is used to classify or predict labels for data points. They are typically trained on a data set of labeled data points. And they learn the relationship between the features of the data points and the labels. Once a discriminative model is trained,it can be used to predict the label for new data points. This type of  model learns the conditional probability distribution or the probability of y, our output, given x, our input, that this is a dog and classifies it as a dog and not a cat.

**Generative Model**

A generative model on the other hand, generates new data instances based on a learned probability distribution of existing data. Thus generative models generate new content. This model learns the joint probability distribution or the probability of x and y and predicts the conditional probability that this is a dog and can then generate a picture of a dog.

# $$y = f(x)$$

- the y stands for the model output
- the f embodies the function used in the calculation
- the x represents the input or inputs used for the formula. 

So, the model output is a function of all inputs. If the y is the number, like predicted sales, it is not gen AI. 

If y is a sentence, like define sales, it is generative as the question would elicit a text response.The response would be based on all the massive large data the model was already trained on.

# 9. How does generative AI really  work? <a class='anchor' id='css'></a> [↑](#top)
The power of generative AI comes from the use of **transformers.** Transformers produced a 2018 revolution in natural language processing. At a high level, a transformer model consists of an encoder and decoder. The encoder encodes the input sequence and passes it to the decoder, which learns how to decode the representation for a relevant task.
​
In transformers, **hallucinations** are words or phrases that are generated by the model that are often nonsensical or grammatically incorrect. 
​
Hallucinations can be caused by a number of factors, 
- The model is not trained on enough data
- The model is trained on noisy or dirty data
- The model is not given enough context
- The model is not given enough constraints. 
​
Hallucinations can be a problem for transformers because they can make the output text difficult to understand. They can also make the model more likely to generate incorrect or misleading information. **a. Text-to-text.**
Text-to-text models take a natural language input and produces a text output. These models are trained to learn the mapping between a pair of text. ### Cool concepts that make  Gen AI even possible

#### **1. Encoder-Decoder Architecture**
We said earlier that Large language models  are a subset of deep learning and intersect with Gen AI  and are both a part of deep learning, encoder-decoder architecture is at the core of large language models.The encoder encodes the input sequence and passes it to the decoder, which learns how to decode the representation for a relevant task, the input could be a prompt.

##### How Does it do it?
The architecture has two stages, the **encoder** stage, which produces a vector representation of the input sentence followed by **decoder**. Encoder and the decoder can be implemented with different internal architectures.They can be a **recurrent neural network (RNN)** or a more complex **transformer** block as in the case of the super powerful language models we see nowadays.During training *greedy* search and *beam search* approaches are used by the decoder to generate the token in your vocabulary that has the highest probability, the latter produces better results.

In large language models the diffrence is what goes inside of the encoder and decoder blocks. Now **attention mechanism** is when the simple RRN network is replaced by **transformer** blocks which is an architecture discovered at Google.

#### **2. Attention Mechanism**
We have said that the encoder encodes the input sequence and passes it to the decoder which learns how to decode the representation for a relevant task like translation, well  to improve the translation the first time step is can add what is called the attention mechanism to the encoder decoder.Attention mechanism is a technique that allows the neural network to focus on specific parts of an input sequence.This is done by assigning weights to different parts of the input sequence with the most important parts receiving the highest weights.

An attention model differs from the traditional sequence to sequence model in two ways.First, the encoder passes a lot more data to the decoder,so the decoder uses all the hidden state information to translate the sentence.The second change that the attention mechanism brings is adding an extra step to the attention decoder before producing its output.


#### **3. Transformers** 
Remember how we said that the power of generative AI comes from the use of transformers.Well, transformers are based on a 2017 paper named [Attention is All You Need](https://research.google/pubs/pub46201/).Transformers produced a 2018 revolution in natural language processingA transformer is an encoder decoder model that uses the attention mechanism. It can take advantage of pluralization and also process a large amount of data at the same time.

Because of its model architecture, attention mechanism helps improve the performance of machine translation applications. Transformer models were built using attention mechanisms at the core. A transformer model consists of encoder and decoder.There's multiple variations of transformers out there now. Some use both the encoder and the decoder component from the original architecture.Some use only the encoder and some use only the decoder. 

#### **4 BERT**
Bert stands for bidirectional encoder representations,a popular encoder only architecture.Bert is one of the trained transformers and was developed by Google in 2018. Since then, multiple variations of Bert have been built. Today, Bert Powers Google Search and is trained on the entire wikipedia and BookCorpus

## Generative AI does present risks, and progress and adoption may slow if these are not considered and mitigated when scaling.

According to the **implications of Generative AI for businesses  2023 report** on a new frontier in Artificial Intelligence by [Deloitte ](https://www2.deloitte.com/us/en/pages/consulting/articles/generative-artificial-intelligence.html), Everyone involved in the development, consumption, discussion,and regulation of Generative AI should strive to manage the following identified risks:

- Erosion of trust 
Malicious—hallucination,deepfakes, phishing, and prompt injection26 —and ambivalent actors—not citing data sources—can expose the attack surface and erode customer trust.

- Security and risk 
Companies should stay ahead of a rapidly evolving regulatory landscape while maintaining confidentiality of data, embeddings, and tuning with inherently “multi-tenant” models.

- Bias and discrimination  
Generative AI is prone to mimicking biases and propagating discriminatory behavior if implemented without guardrails and continuous monitoring.

- Data privacy and IP obscurity 
Models will be trained on a corpus of proprietary, often private data requiring regulatory compliance, node isolation, and source traceability.

- Costs 
Costs of a query/prompt using Generative AI can cost up to ten times27 that of an index-based query. While these costs will likely come down over time, the economics should be factored into internal business cases and customer pricing to drive adoption.

- Long-term worker displacement 
Today, the highest ROI use cases will augment workflows and drive productivity; however,as models advance, there may be a risk of job displacement without proper upskilling and workforce planning.
